```python
import pandas as pd

# Load the dataset
df = pd.read_excel('Online Retail.xlsx')

# Inspect the first 5 rows and data types
print(df.head())
print(df.info())
```

```python
# Drop rows with missing values in critical columns
df = df.dropna(subset=['InvoiceNo', 'Description'])

# Remove cancelled orders (typically denoted by 'C' in invoice numbers)
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]

# Ensure quantity is positive
df = df[df['Quantity'] > 0]

print(f"Cleaned dataset shape: {df.shape}")
```

```python
from mlxtend.frequent_patterns import apriori, association_rules

# Create a basket matrix (Transactions as rows, Products as columns)
basket = (df.groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0)
          .set_index('InvoiceNo'))

# Convert quantities to binary presence (1 for purchased, 0 otherwise)
def encode_units(x):
    return 1 if x > 0 else 0

basket_sets = basket.applymap(encode_units)

# Generate frequent itemsets (adjust min_support based on dataset size)
frequent_itemsets = apriori(basket_sets, min_support=0.01, use_colnames=True)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)

# Display top 5 strongest rules sorted by confidence and lift
print(rules.sort_values(['confidence', 'lift'], ascending=[False, False]).head())
```
